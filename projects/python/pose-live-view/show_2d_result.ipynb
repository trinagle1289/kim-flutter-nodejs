##### 主程式

```python
# view result
import cv2

# data process
import numpy as np
import time

# datatype
from mediapipe.tasks.python.components.containers.landmark import Landmark

# custom
from src.mediapipe_lib.base import PoseLandmarkerLiveStream
```

```python
lanmarker_live = PoseLandmarkerLiveStream(MODEL)
cap = cv2.VideoCapture(0)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break

    # 運算模型
    lanmarker_live.detect_async(frame, int(time.time() * 1000))

    # 模型運算結果(會實時更新)
    mp_img = lanmarker_live.current_image
    cur_result = lanmarker_live.result

    if mp_img is None:  # 無法取得 mp_img
        continue

    # 轉換 mediapipe 圖片為 numpy 格式
    np_img = np.array(mp_img.numpy_view())

    if len(cur_result.pose_landmarks) <= 0:  # 無法取得身體姿勢結果
        cv2.imshow("result", np_img)
        if cv2.waitKey(1) == ord("q"):  # 按下 Q 離開
            break
        else:
            continue

    # 原圖繪圖
    h, w, c = np_img.shape
    for kpt in cur_result.pose_landmarks[0]:
        kpt: Landmark
        pos = (int(kpt.x * w), int(kpt.y * h))
        cv2.circle(np_img, pos, 5, (0, 0, 255), -1)

    cv2.imshow("result", np_img)  # 顯示繪圖結果

    if cv2.waitKey(1) == ord("q"):  # 按下 Q 離開
        break

# 關閉物件
cap.release()
cv2.destroyAllWindows()
lanmarker_live.close()
```
