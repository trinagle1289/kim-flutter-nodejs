```python
def deduplicate_strings(strings):
    if not strings:
        return []
    
    # 初始情況：保留第一個字串
    deduped = [strings[0]]
    
    for i in range(1, len(strings)):
        # 如果當前字串與上一次保留的字串不同，則添加到結果中
        if strings[i] != deduped[-1]:
            deduped.append(strings[i])
    
    return deduped

# 測試數據
input_strings = ["a", "a", "b", "b", "b", "c", "a", "a", "d", "d", "c", "c"]
output_strings = deduplicate_strings(input_strings)
print("去抖動後的字串陣列:", output_strings)
```
