```python
def smooth_array(arr):
    smoothed_arr = []
    count = 1
    n = len(arr)

    for i in range(1, n):
        if arr[i] == arr[i - 1]:
            count += 1
        else:
            if count > 15:
                smoothed_arr.extend([arr[i - 1]] * count)
            count = 1

    # Check the last segment
    if count > 15:
        smoothed_arr.extend([arr[-1]] * count)

    return smoothed_arr

# 示例數據
original_array = ["A1"] * 20 + ["A2"] * 18 + ["A3"] * 25 + ["A4"] * 16 + ["A5"] * 10

# 平滑數據
smoothed_array = smooth_array(original_array)
print(smoothed_array)

print(len(original_array))
print(len(smoothed_array))
```
