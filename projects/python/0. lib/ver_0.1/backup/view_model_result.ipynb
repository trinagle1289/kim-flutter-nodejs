#### 串流分析

```python
landmarker_live = PoseLandmarkerLiveStream(MODEL)
cap = cv2.VideoCapture(0)

while cap.isOpened():
    # 取得影片幀
    ret, frame = cap.read()
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")

    # 進行姿勢偵測
    landmarker_live.detect_async(frame, int(time.time() * 1000))
    live_result = landmarker_live.result
    live_image: mp.Image = landmarker_live.current_image
    live_timestamp_ms = landmarker_live.current_timestamp_ms

    # 按下 Q 離開
    if cv2.waitKey(1) == ord("q"):
        break
    # 如果結果為空則直接跳下一幀
    if live_image is None:
        continue
    # 抓取不到姿勢直接跳下一幀
    if len(live_result.pose_landmarks) <= 0:
        cv2.imshow("result", live_image.numpy_view())
        continue

    # 繪製圖片
    np_img = np.array(live_image.numpy_view())
    h, w, c = np_img.shape
    for kpt in live_result.pose_landmarks[0]:
        pos = (int(kpt.x * w), int(kpt.y * h))
        np_img = cv2.circle(np_img, pos, 5, (0, 0, 255), -1)

    # 顯示圖片
    cv2.imshow("result", np_img)

# 關閉所有物件
landmarker_live.close()
cap.release()
cv2.destroyAllWindows()
print("finished")
```
