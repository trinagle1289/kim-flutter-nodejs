```python
IPYNB_FOLDER = "./src_copy/"
OUTPUT_FOLDER = "./output/"
```

```python
from pathlib import Path
import subprocess
```

```python
folder_in = Path(IPYNB_FOLDER)
folder_out = Path(OUTPUT_FOLDER)
```

```python
result = subprocess.run(
    "echo 123",
    stdout=subprocess.PIPE,
    shell=True,
    text=True,
)
```

```python
for pth in list(folder_in.rglob("*.ipynb")):
    saved_folder = folder_out / pth.parent
    cmd = f"jupyter nbconvert --output-dir='{saved_folder}' --to python {pth}"
    result = subprocess.run(cmd, shell=True, text=True)
    print(f"finish cmd: {result.args}")

print("finish")
```
