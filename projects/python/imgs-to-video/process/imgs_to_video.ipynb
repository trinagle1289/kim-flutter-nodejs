```python
IMG_IN_PATH = "../../../resources/result pics/A1-A3/BlazePose TFjs/"
VIDEO_OUT_PATH = "test.mp4"
VIDEO_FRAME_RATE = 30
VIDEO_SIZE = (1080, 1920)
```

```python
import os
import cv2
```

```python
frames: list[cv2.typing.MatLike] = []
```

```python
# 將圖片存入到陣列中
for filename in os.listdir(IMG_IN_PATH):
    img_path = os.path.join(IMG_IN_PATH, filename)
    img = cv2.imread(img_path)
    frames.append(img)
```

```python
# 設定影像寫入物件
fourcc = cv2.VideoWriter.fourcc(*"h264")
video_out = cv2.VideoWriter(VIDEO_OUT_PATH, fourcc, 30, VIDEO_SIZE)
```

```python
# 將陣列中的圖片存入到影像中
for f in frames:
    video_out.write(f)

video_out.release()
```
