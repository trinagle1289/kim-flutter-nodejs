```python
POSE_TAG = "A4_2-A2"
JSON_FILE_PATH = f"./json/front/{POSE_TAG}.json"
OUTPUT_FOLDER_PATH = f"./output/front/angle_left/{POSE_TAG}/"
```

```python
from src.json_pose import PoseData3D
import src.kpt_special_data as kpt_special_data
import plot_painter as plot_painter
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
data = PoseData3D(JSON_FILE_PATH)
json_pth = Path(JSON_FILE_PATH)
con_lst = kpt_special_data.blazepose_connection
out_pth = Path(OUTPUT_FOLDER_PATH)
```

```python
# 創建輸出資料夾
if not out_pth.exists():
    out_pth.mkdir(parents=True)
```

```python
# 參數設定
data_range = [-1, 1]  # 顯示資料範圍
elev, azim, roll = 20, 30, 0  # 設定視角
is_3d = True  # 圖表是否為 3D
c_left = "#f00"  # 左邊身體顏色
c_center = "#0f0"  # 右邊身體顏色
c_right = "#00f"  # 中間身體顏色
c_dot = "#000"  # 關鍵點顏色
dot_size = 5  # 繪製關鍵點大小

for i in range(data.get_number_of_frames()):
    title_name = ""  # 標題名稱

    # 圖表設置
    fig = plt.figure()
    ax = plt.subplot(projection="3d") if is_3d else plt.subplot()
    ax.view_init(elev, azim, roll)  # 設置視角
    plot_painter.set_data_range(data_range, ax)

    if data.get_number_of_poses(i) > 0:  # 如果可以讀取到姿勢
        # 設定資料
        title_name = (
            f"label:{data.get_lhc_label(i, 0)}\nscore:{data.get_kpt_score(i, 0)}"
        )
        p_kpt = data.get_pos_list(i, 0, is_3d)
        p_left, p_center, p_right = data.get_pos_connection_list(i, 0, con_lst, is_3d)
        # 繪製圖表
        plot_painter.draw_lines(p_left, is_3d, c_left, ax)
        plot_painter.draw_lines(p_center, is_3d, c_center, ax)
        plot_painter.draw_lines(p_right, is_3d, c_right, ax)
        plot_painter.draw_dots(p_kpt, is_3d, dot_size, c_dot, ax)
    else:  # 如果讀取不到姿勢
        title_name = "No pose"

    # 顯示圖表
    ax.set_title(title_name)
    plt.savefig(f"{OUTPUT_FOLDER_PATH}{POSE_TAG}({i:03d}).png")
    plt.close()

print("Finish")
```
