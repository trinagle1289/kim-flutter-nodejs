```python
POSE_TAG = "A1-A3"
JSON_FILE_PATH = f"./json/front/{POSE_TAG}.json"
OUTPUT_FOLDER_PATH = f"./output/front/{POSE_TAG}/"
```

```python
from src.json_pose import PoseData3D
import src.kpt_special_data as kpt_special_data
import src.draw_plot as draw_plot
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
data = PoseData3D(JSON_FILE_PATH)
json_pth = Path(JSON_FILE_PATH)
con_lst = kpt_special_data.blazepose_connection
out_pth = Path(OUTPUT_FOLDER_PATH)
```

```python
# 創建輸出資料夾
if not out_pth.exists():
    out_pth.mkdir(parents=True)
```

```python
for i in range(data.get_number_of_frames()):
    title_name = f"label: {data.get_lhc_label(i, 0)}\nscore: {data.get_kpt_score(i, 0)}"  # 標題名稱

    if data.get_number_of_poses(i) <= 0:  # 錯誤判斷(讀取不到姿勢問題)
        title_name = "No pose"
        continue

    # 設定資料
    pos = data.get_pos_list(i, 0, is_3d=True)
    left, center, right = data.get_pos_connection_list(i, 0, con_lst, is_3d=True)

    # 建立 3D 圖表
    fig = plt.figure()
    ax = plt.subplot(projection="3d")
    ax.set_title(title_name)
    ax.view_init(elev=20, azim=30, roll=0)  # 設置視角
    ax = draw_plot.set_plot_data_range([-1, 1], ax)

    # 繪製圖表
    ax = draw_plot.get_line_plot(left, is_3d=True, color="#f00", ax=ax)
    ax = draw_plot.get_line_plot(center, is_3d=True, color="#0f0", ax=ax)
    ax = draw_plot.get_line_plot(right, is_3d=True, color="#00f", ax=ax)
    ax = draw_plot.get_dots_in_scatter(pos, is_3d=True, color="#000", ax=ax)

    # 顯示圖表
    plt.savefig(f"{OUTPUT_FOLDER_PATH}{POSE_TAG}({i:03d}).png")
    plt.close()

print("Finish")
```
