```python
POSE_TAG = "A1-A3"
JSON_FILE_PATH = f"./json/{POSE_TAG}.json"
OUTPUT_FOLDER_PATH = f"./output/{POSE_TAG}/"
```

```python
from src.json_pose import PoseData3D
import src.kpt_special_data as kpt_special_data
import src.draw_plot as draw_plot
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
data = PoseData3D(JSON_FILE_PATH)
con_lst = kpt_special_data.blazepose_connection
out_pth = Path(OUTPUT_FOLDER_PATH)
```

```python
# 創建輸出資料夾
if not out_pth.exists():
    out_pth.mkdir(parents=True)
```

```python
lables = data.to_label_list()

for i in range(data.get_number_of_frames()):
    # 設定資料
    pos = data.get_pos_list(i, 0, is_3d=True)
    left_con, center_con, right_con = data.get_pos_connection_list(
        i, 0, con_lst, is_3d=True
    )

    # 建立 3D 圖表
    fig = plt.figure()
    ax = plt.subplot(projection="3d")
    ax.set_title(lables[i])
    # 繪製線條
    ax = draw_plot.get_line_plot(
        left_con, data_range=[-1, 1], is_3d=True, color="#f00", ax=ax
    )
    ax = draw_plot.get_line_plot(center_con, is_3d=True, color="#0f0", ax=ax)
    ax = draw_plot.get_line_plot(right_con, is_3d=True, color="#00f", ax=ax)
    # 繪製座標點
    ax = draw_plot.get_dots_in_scatter(pos, is_3d=True, color="#000", ax=ax)
    # 設置視角
    ax.view_init(elev=20, azim=30, roll=0)

    # 顯示圖表
    plt.savefig(f"{OUTPUT_FOLDER_PATH}{POSE_TAG}({i:03d}).png")
    plt.close()

print("Finish")
```
