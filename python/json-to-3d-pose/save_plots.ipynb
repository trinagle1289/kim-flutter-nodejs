```python
POSE_TAG = "A5_3-A3"
JSON_FILE_PATH = f"./json/side/{POSE_TAG}.json"
OUTPUT_FOLDER_PATH = f"./output/side/angle_front/{POSE_TAG}/"
```

```python
from src.json_pose import PoseData3D
import src.kpt_special_data as kpt_special_data
import src.plot_painter as plot_painter
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
```

```python
### 視角參數

### 預設
# elev, azim, roll = 20, 30, 0

## 正面影像
# elev, azim, roll = 30, 90, 0  # 正面
# elev, azim, roll = 30, 180, 0  # 右側

## 右側影像
# elev, azim, roll = 30, 0, 0  # 正面
# elev, azim, roll = 30, 270, 0  # 右側
```

##### 正面圖片輸出

```python
### 進行資料的建立
data = PoseData3D(JSON_FILE_PATH)  # 取得 json 檔資料
json_pth = Path(JSON_FILE_PATH)  # json 檔案路徑
out_pth = Path(OUTPUT_FOLDER_PATH)  # 輸出檔案路徑

# 創建輸出資料夾
if not out_pth.exists():
    out_pth.mkdir(parents=True)
```

```python
### 參數設定

# 設定視角
elev, azim, roll = 30, 0, 0  # 正面

# 顯示資料範圍
data_range_x = [-1, 1]  # x 軸
data_range_y = [-1, 1]  # y 軸
data_range_z = [0, 2]  # z 軸

# 圖表資訊顏色
c_left = "#f00"  # 左邊
c_center = "#0f0"  # 右邊
c_right = "#00f"  # 中間
c_dot = "#000"  # 關鍵點

is_3d = True  # 圖表是否為 3D
dot_size = 5  # 繪製關鍵點大小
con_lst = kpt_special_data.blazepose_connection  # 設定連線列表
output_file_prefix = f"{OUTPUT_FOLDER_PATH}{POSE_TAG}"  # 輸出檔案前綴
```

```python
### 對全部影像幀進行分析
for i in range(data.get_number_of_frames()):
    title_name = ""  # 標題名稱

    ### 圖表設置
    fig = plt.figure()
    ax = plt.subplot(projection="3d") if is_3d else plt.subplot()
    ax.view_init(elev, azim, roll)  # 設置視角
    # 設置參數範圍
    plot_painter.set_data_range(data_range_x, data_range_y, data_range_z, ax)

    if data.get_number_of_poses(i) > 0:  # 如果可以讀取到姿勢
        ### 取得資料
        title_name = (
            f"label:{data.get_lhc_label(i, 0)}\nscore:{data.get_kpt_score(i, 0)}"
        )
        p_kpt = data.get_pos_list(i, 0, is_3d)
        p_left, p_center, p_right = data.get_pos_connection_list(i, 0, con_lst, is_3d)

        ### 轉換成 numpy 格式
        p_left = np.array(p_left)
        p_center = np.array(p_center)
        p_right = np.array(p_right)
        p_kpt = np.array(p_kpt)
        # 取得 y 軸要加的數值
        plus_to_y = p_kpt[:, 1].max()
        # 反轉 y 軸數值，並加上要加的數值(腳底為0)
        p_left[:, :, 1] = p_left[:, :, 1] * -1 + plus_to_y
        p_center[:, :, 1] = p_center[:, :, 1] * -1 + plus_to_y
        p_right[:, :, 1] = p_right[:, :, 1] * -1 + plus_to_y
        p_kpt[:, 1] = p_kpt[:, 1] * -1 + plus_to_y

        ### 繪製圖表
        plot_painter.draw_lines(p_left, is_3d, c_left, ax)
        plot_painter.draw_lines(p_center, is_3d, c_center, ax)
        plot_painter.draw_lines(p_right, is_3d, c_right, ax)
        plot_painter.draw_dots(p_kpt, is_3d, dot_size, c_dot, ax)
    else:  # 如果讀取不到姿勢
        title_name = "No pose"

    ### 圖表輸出
    ax.set_title(title_name)
    plt.savefig(f"{output_file_prefix}({i:03d}).png")
    plt.close()
```
