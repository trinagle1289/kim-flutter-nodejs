```python
import matplotlib.pyplot as plt
```

```python
def get_pose_line_chart(
    labels: list[str], data_type: str = None, ax: plt.Axes = None
) -> plt.Axes:
    """
    取得姿勢結果折線圖
    labels: 姿勢標籤列表
    data_type: 所代表的模型
    ax: Matplotlib 座標資料
    """
    POSE_LABEL = ["A1", "A2", "A3", "A4", "A5"]
    data = []

    # 將標籤以數值的形式存入到 data 中
    for lab in labels:
        if lab == "A1":
            data.append(0)
        if lab == "A2":
            data.append(1)
        if lab == "A3":
            data.append(2)
        if lab == "A4":
            data.append(3)
        if lab == "A5":
            data.append(4)

    if ax is None:
        ax = plt.gca()

    ax.set_xlabel("time frame")
    ax.set_ylabel("pose label")
    ax.set_yticks(list(range(5)), POSE_LABEL)
    ax.plot(range(len(data)), data, label=data_type)
    return ax
```
