## Mediapipe 轉換成 Pose 物件測試

```python
import mediapipe as mp
import cv2

from mediapipe.tasks.python.core.base_options import BaseOptions
from mediapipe.tasks.python.vision.core.vision_task_running_mode import (
    VisionTaskRunningMode,
)
from mediapipe.tasks.python.vision.pose_landmarker import (
    PoseLandmarker,
    PoseLandmarkerOptions,
)
```

##### 忽略警告

```python
import warnings

warnings.filterwarnings("ignore")
```

##### 主程式

```python
MODEL = "./models/BlazePose/pose_landmarker_lite.task"
IMG = "./images/pexels-photo-4384679.jpeg"
```

```python
options = PoseLandmarkerOptions(
    base_options=BaseOptions(model_asset_path=MODEL),
    running_mode=VisionTaskRunningMode.IMAGE,
)
```

```python
cv_mat = cv2.imread(IMG)
w, h, c = cv_mat.shape
cv_mat = cv2.resize(cv_mat, (h // 5, w // 5))

frame = mp.Image(image_format=mp.ImageFormat.SRGB, data=cv_mat)

with PoseLandmarker.create_from_options(options) as landmarker:
    result = landmarker.detect(frame)
```

```python
out_mat = cv_mat.copy()
h, w, c = out_mat.shape  # 取得圖片資訊

for kpt in result.pose_landmarks[0]:
    pos = (int(kpt.x * w), int(kpt.y * h))  # 計算座標
    cv2.circle(out_mat, pos, 3, (0, 0, 255), -1)  # 繪製圓圈

cv2.imshow("result", out_mat)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
