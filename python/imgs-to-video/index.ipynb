```python
# dir1 為 INPUT_PATH/ 底下個別的資料夾
# out_dir1 為 OUTPUT_PATH/ 底下個別的資料夾
# dir2 為 INPUT_PATH/dir1/ 底下個別的資料夾
# out_dir2 為 OUTPUT_PATH/out_dir1/ 底下個別的資料夾
# dir3 為 INPUT_PATH/dir1/dir2/ 底下個別的資料夾
# vid_out_pth 為 OUTPUT_PATH/out_dir2/ 底下個別的影片路徑
for dir1 in get_folders_in_folder(RESULT_PATH):
    out_dir1 = Path(OUTPUT_PATH) / dir1.name
    if not out_dir1.exists():
        out_dir1.mkdir(parents=True)

    for dir2 in get_folders_in_folder(dir1):
        out_dir2 = out_dir1 / dir2.name
        if not out_dir2.exists():
            out_dir2.mkdir(parents=True)

        for dir3 in get_folders_in_folder(dir2):
            vid_out_pth = out_dir2 / (dir3.name + ".mp4")  # 影片檔名
            img_folder_to_video(str(dir3), str(vid_out_pth), (640, 480), 10)
            print(f"Finish Creating video in {vid_out_pth}")
```
