```python
RESULT_PATH = "../../resources/results/"
OUTPUT_PATH = "./tmp-output/"
```

```python
from src.utils import get_folders_in_folder, img_folder_to_video
from pathlib import Path
```

```python
# dir1 為 INPUT_PATH 底下個別的資料夾
# out_dir1 為 OUTPUT_PATH 底下個別的資料夾
# dir2 為 INPUT_PATH/dir1 底下個別的資料夾
# vid_out_pth 為 OUTPUT_PATH/out_dir2 底下個別的影片路徑
for dir1 in get_folders_in_folder(RESULT_PATH):
    out_dir1 = Path(OUTPUT_PATH) / dir1.name  # 影片輸出資料夾
    # 建立影片輸出資料夾
    if not out_dir1.exists():
        out_dir1.mkdir(parents=True)

    for dir2 in get_folders_in_folder(dir1):
        vid_out_pth = out_dir1 / (dir2.name + ".mp4")  # 影片檔名
        img_folder_to_video(str(dir2), str(vid_out_pth), (1080, 1920))
        print(f"Finish Creating video in {vid_out_pth}")
```
