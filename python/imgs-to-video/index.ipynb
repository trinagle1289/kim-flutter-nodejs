```python
RESULT_PATH = "../../resources/results/"
OUTPUT_PATH = "./output/"
```

```python
from src.utils import get_folders_in_folder, img_folder_to_video
from pathlib import Path
```

```python
for pose_dir in get_folders_in_folder(RESULT_PATH):
    video_dir = Path(OUTPUT_PATH) / pose_dir.name  # 影片輸出資料夾

    if not video_dir.exists():
        video_dir.mkdir(parents=True)

    for model_dir in get_folders_in_folder(pose_dir):
        video_path = video_dir / (model_dir.name + ".mp4")  # 影片檔名
        img_folder_to_video(str(model_dir), str(video_path), (1080, 1920))
        print(f"Finish Creating video in {video_path}")
```
